```python
# import all the lib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# read the dataset using pandas
data = pd.read_csv('Salary_Data.csv')
```

```python
# This displays the top 5 rows of the data
data.head()
```

```python
# Provides some information regarding the columns in the data
data.info()
```

```python
# this describes the basic stat behind the dataset used 
data.describe()
```

```python
# These Plots help to explain the values and how they are scattered

plt.figure(figsize=(12,6))
sns.pairplot(data,x_vars=['YearsExperience'],y_vars=['Salary'],height=7,kind='scatter')
plt.xlabel('Years')
plt.ylabel('Salary')
plt.title('Salary Prediction')
plt.show()
```

```python
# Cooking the data
X = data['YearsExperience']
X.head()
```

```python
# Cooking the data
y = data['Salary']
y.head()
```

```python
# Import Segregating data from scikit learn
from sklearn.model_selection import train_test_split
```

```python
# Split the data for train and test 
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.7,random_state=100)
```

```python
# Create new axis for x column
X_train = X_train[:,np.newaxis]
X_test = X_test[:,np.newaxis]
```

```python
# Importing Linear Regression model from scikit learn
from sklearn.linear_model import LinearRegression
```

```python
# Fitting the model
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
# Predicting the Salary for the Test values
y_pred = lr.predict(X_test)
```

```python
# Plotting the actual and predicted values

plt.scatter(X_test,y_test,color='r')
plt.plot(X_test,y_pred,color='b',linestyle='-')
plt.xlabel('Years')
plt.ylabel('Salary')
plt.title('Prediction')
plt.show()
```

```python
# Importing metrics for the evaluation of the model
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
# calculate Mean square and absolute error
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
```

```python
# Calculate R square value by checking the difference between the actual test data and the predicted value
rsq = r2_score(y_test,y_pred)
```

```python
print('mean squared error :',mse)
print('mean absolute error :',mae)
print('r square :',rsq)
```

```python
# Intecept and coeff of the line
print('Intercept of the model:',lr.intercept_)
print('Coefficient of the line:',lr.coef_)
```
