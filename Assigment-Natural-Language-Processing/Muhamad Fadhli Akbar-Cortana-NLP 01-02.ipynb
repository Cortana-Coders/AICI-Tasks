## Scrape From Multiple Link

```python
def url_to_transcript(url):
    page = requests.get(url).text
    soup = BeautifulSoup(page, 'lxml')
    text = [p.text for p in soup.find_all('div', class_="elementor-widget-theme-post-content")]
    print(url)
    return text
```

```python
# URLs of transcripts in scope
urls = ['https://scrapsfromtheloft.com/movies/ghostbusters-afterlife-transcript/',
        'https://scrapsfromtheloft.com/movies/black-panther-wakanda-forever-transcript/',
        'https://scrapsfromtheloft.com/movies/shazam-fury-of-the-gods-transcript/']
```

```python
transcript = [url_to_transcript(u) for u in urls]
```
