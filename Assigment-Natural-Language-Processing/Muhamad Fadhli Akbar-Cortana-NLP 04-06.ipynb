````python
import numpy as np
import pandas as pd
! pip install textblob
````

````python
from textblob import TextBlob
````

````python
df = pd.read_csv('IMDB_Dataset.csv')
````

````python
! pip install vaderSentiment
````

````python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
````

````python
def get_polarity(text):
    return TextBlob(str(text)).sentiment.polarity

# Function to determine sentiment based on polarity
def get_sentiment_from_polarity(polarity):
    if polarity > 0:
        return 'positive'
    elif polarity < 0:
        return 'negative'
    else:
        return 'neutral'
````

````python
def get_vader_sentiment(text):
    analyzer = SentimentIntensityAnalyzer()
    vs = analyzer.polarity_scores(str(text))
    if vs['compound'] >= 0.05:
        return 'positive'
    elif vs['compound'] <= -0.05:
        return 'negative'
    else:
        return 'neutral'
````

````python
# Calculate polarity and sentiment using TextBlob
df['polarity'] = df['review'].apply(get_polarity)
df['sentiment_polarity'] = df['polarity'].apply(get_sentiment_from_polarity)

# Calculate sentiment using VADER
df['sentiment_vader'] = df['review'].apply(get_vader_sentiment)

# Compare with ground truth
df['match_polarity'] = df['sentiment_polarity'] == df['sentiment']
df['match_vader'] = df['sentiment_vader'] == df['sentiment']
````

````python
accuracy_polarity = (df['match_polarity'].sum() / len(df)) * 100
accuracy_vader = (df['match_vader'].sum() / len(df)) * 100
````

````python
print("=== Sentiment Analysis Results ===\n")
print("TextBlob Polarity-based Analysis:")
print(f"Accuracy: {accuracy_polarity:.2f}%")
print("\nSentiment Distribution:")
print(df['sentiment_polarity'].value_counts())
print("\n" + "="*30)
````

````python
print("\nVADER Analysis:")
print(f"Accuracy: {accuracy_vader:.2f}%")
print("\nSentiment Distribution:")
print(df['sentiment_vader'].value_counts())
````

````python
# Sample comparison (first 5 reviews)
print("\n=== Sample Comparison (First 5 Reviews) ===")
comparison_df = df[['review', 'sentiment', 'sentiment_polarity', 'sentiment_vader']].head()
print(comparison_df)
````
