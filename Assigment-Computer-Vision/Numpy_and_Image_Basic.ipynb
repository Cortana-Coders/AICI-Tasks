# Images and Numpy

## Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Import the data

```python
from PIL import Image
```

```python
pic = Image.open('741500.jpg')
```

```python
pic
```

```python
type(pic)
```

```python
pic_arr = np.asarray(pic)
pic_arr.shape
```

```python
plt.imshow(pic_arr)
```

```python
pic_red = pic_arr.copy()
pic_green = pic_arr.copy()
pic_blue = pic_arr.copy()
```

```python
# Show red picture
pic_red[:, :, 1] = 0
pic_red[:, :, 2] = 0

plt.imshow(pic_red)
```

```python
# Show green picture
pic_green[:, :, 0] = 0
pic_green[:, :, 2] = 0

plt.imshow(pic_green)
```

```python
# Show blue picture
pic_blue[:, :, 0] = 0
pic_blue[:, :, 1] = 0

plt.imshow(pic_blue)
```
