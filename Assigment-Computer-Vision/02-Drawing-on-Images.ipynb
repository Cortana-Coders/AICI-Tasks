## Shapes

### Rectangles

* img image.
* pt1 Vertex of the rectangle.
* pt2 Vertex of the rectangle opposite to pt1 .
* color Rectangle color or brightness (grayscale image).
* thickness Thickness of lines that make up the rectangle. Negative values, like #FILLED,mean that the function has to draw a filled rectangle.
* lineType Type of the line. See #LineTypes
* shift Number of fractional bits in the point coordinates.

```python
# pt1 = top left
# pt2 = bottom right
cv2.rectangle(blank_img,pt1=(384,0),pt2=(510,128),color=(0,255,0),thickness=5)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a blank image
serius_wir = np.zeros((800, 800, 3), dtype=np.uint8)

# Draw the left circle
cv2.circle(img=serius_wir, center=(300, 300), radius=100, color=(255, 0, 0), thickness=-1)

# Draw the right circle
cv2.circle(img=serius_wir, center=(500, 300), radius=100, color=(255, 0, 0), thickness=-1)

# Draw the bottom part of the heart using a polygon
points = np.array([[200, 300], [400, 600], [600, 300], [500, 500], [300, 500]], np.int32)
points = points.reshape((-1, 1, 2))
cv2.fillPoly(serius_wir, [points], color=(255, 0, 0))

# Display the image
plt.imshow(serius_wir)
plt.axis('off')  # Hide the axes
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
sample_dog_cv = cv2.imread('doggo.jpg')

# Ensure the image is loaded
if sample_dog_cv is None:
    raise FileNotFoundError("Image file 'doggo.jpg' not found.")

font = cv2.FONT_HERSHEY_SIMPLEX
# Koordinat (x,y) dan (x,y) adalah koordinat titik awal dari text
cv2.putText(sample_dog_cv,text='I',org=(25,40),fontFace=font,fontScale= 1,color=(255,255,255),thickness=1,lineType=cv2.LINE_AA)
plt.imshow(sample_dog_cv)

# Koordinat (x,y) dan (x,y) adalah koordinat titik awal dari text
cv2.putText(sample_dog_cv,text='Dog',org=(10,100),fontFace=font,fontScale= 0.8,color=(255,255,255),thickness=1,lineType=cv2.LINE_AA)
plt.imshow(sample_dog_cv)

# Get the dimensions of the image
height, width, _ = sample_dog_cv.shape

# Define the center and size of the heart
center_x, center_y = width // 6, height // 2
heart_size = min(width, height) // 7

# Draw the left circle
cv2.circle(img=sample_dog_cv, center=(center_x - heart_size // 2, center_y - heart_size // 4), radius=heart_size // 2, color=(255, 0, 0), thickness=-1)

# Draw the right circle
cv2.circle(img=sample_dog_cv, center=(center_x + heart_size // 2, center_y - heart_size // 4), radius=heart_size // 2, color=(255, 0, 0), thickness=-1)

# Draw the bottom part of the heart using a polygon
points = np.array([
    [center_x - heart_size, center_y - heart_size // 6],
    [center_x + heart_size, center_y - heart_size // 6],
    [center_x, center_y + heart_size]
], np.int32)
points = points.reshape((-1, 1, 2))
cv2.fillPoly(sample_dog_cv, [points], color=(255, 0, 0))

# Display the image
plt.imshow(cv2.cvtColor(sample_dog_cv, cv2.COLOR_BGR2RGB))
plt.axis('off')  # Hide the axes
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
sample_dog_cv = cv2.imread('doggo.jpg')

# Ensure the image is loaded
if sample_dog_cv is None:
    raise FileNotFoundError("Image file 'doggo.jpg' not found.")

# Define the size and position of the heart
heart_width = 50
heart_height = 70
center_x, center_y = 100, 100  # Position of the heart

# Draw the left circle
cv2.circle(img=sample_dog_cv, center=(center_x - heart_width // 4, center_y - heart_height // 4), radius=heart_width // 4, color=(255, 0, 0), thickness=-1)

# Draw the right circle
cv2.circle(img=sample_dog_cv, center=(center_x + heart_width // 4, center_y - heart_height // 4), radius=heart_width // 4, color=(255, 0, 0), thickness=-1)

# Draw the bottom part of the heart using a polygon
points = np.array([
    [center_x - heart_width // 2, center_y - heart_height // 4],
    [center_x + heart_width // 2, center_y - heart_height // 4],
    [center_x, center_y + heart_height // 2]
], np.int32)
points = points.reshape((-1, 1, 2))
cv2.fillPoly(sample_dog_cv, [points], color=(255, 0, 0))

# Add text to the image
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(sample_dog_cv, text='Dog', org=(10, 100), fontFace=font, fontScale=0.8, color=(255, 255, 255), thickness=1, lineType=cv2.LINE_AA)

# Display the image
plt.imshow(cv2.cvtColor(sample_dog_cv, cv2.COLOR_BGR2RGB))
plt.axis('off')  # Hide the axes
plt.show()
```
