```python
sample_dog_cv = cv2.imread('doggo.jpg')
```

```python
plt.imshow(sample_dog_cv)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
# Koordinat (x,y) dan (x,y) adalah koordinat titik awal dari text
cv2.putText(sample_dog_cv,text='I',org=(25,40),fontFace=font,fontScale= 1,color=(255,255,255),thickness=1,lineType=cv2.LINE_AA)
plt.imshow(sample_dog_cv)
```

```python
# Koordinat (x,y) dan (x,y) adalah koordinat titik awal dari text
cv2.putText(sample_dog_cv,text='Dog',org=(10,100),fontFace=font,fontScale= 0.8,color=(255,255,255),thickness=1,lineType=cv2.LINE_AA)
plt.imshow(sample_dog_cv)
```

```python
height, width, _ = sample_dog_cv.shape

# Define the center and size of the heart
center_x, center_y = width // 6, height // 2
heart_size = min(width, height) // 7

# Draw the left circle
cv2.circle(img=sample_dog_cv, center=(center_x - heart_size // 2, center_y - heart_size // 4), radius=heart_size // 2, color=(255, 0, 0), thickness=-1)

# Draw the right circle
cv2.circle(img=sample_dog_cv, center=(center_x + heart_size // 2, center_y - heart_size // 4), radius=heart_size // 2, color=(255, 0, 0), thickness=-1)

# Draw the bottom part of the heart using a polygon
points = np.array([
    [center_x - heart_size, center_y - heart_size // 6],
    [center_x + heart_size, center_y - heart_size // 6],
    [center_x, center_y + heart_size]
], np.int32)
points = points.reshape((-1, 1, 2))
cv2.fillPoly(sample_dog_cv, [points], color=(255, 0, 0))

# Display the image
plt.imshow(sample_dog_cv)
plt.axis('off')  # Hide the axes
plt.show()
```
